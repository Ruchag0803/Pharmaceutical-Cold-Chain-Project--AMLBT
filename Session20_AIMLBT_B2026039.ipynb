Parameter: Internal variables whose values are learned automatically from the training data during model training. Optimized by the training algorithm (e.g., via Gradient Descent). Weights ($w$) and biases ($b$) in a neural network; coefficients in linear regression.

Hyperparameter: External configuration settings whose values are set manually before training begins. Optimized externally through trial-and-error, grid search, or automated tuning. Learning rate, number of layers, batch size, tree depth in a random forest.
